```python
#the groupby() function creates a pandas object (indicated by the term “object” in the output message above)
df.groupby('product_name')
```

```python
#Apply the agg() function to each group to obtain the mean values for the “order_number” column


df.groupby('department_id').agg({'order_number': ['mean']})
```

#groupby() function is being assigned to the df dataframe
Onto this resulting 'object', the agg() function 'is' applied. 
The agg() function will return the mean of the given column

```python
df.groupby('department_id')['order_number'].mean()
```

when using agg(), put the column you want to aggregate inside the parentheses of the agg() function as an argument. When using mean() (or any other standard aggregation function), simply index the column with square brackets, then follow it with the function you want to use after the dot

```python
#Performing Multiple Aggregations
```

```python
df.groupby('department_id').agg({'order_number': ['mean', 'min', 'max']})
```

```python
df_prods_ords_merge['max_order'] = df_prods_ords_merge.groupby(['user_id'])['order_number'].transform(np.max)
```

 -new column called “max_order” 
 
 -the ords_prods_merge dataframe is grouped by the “user_id” column 
 
 -the transform() function is applied on the “order_number” column with the np.max argument
