```python
#Creating Bar Charts

Creating a bar chart in a df follows a similar procedure to that of printing the frequency for a specific column—the only
difference being that the result will be visual rather than just a table of values. 

syntax:
'ords_prods_merge['orders_day_of_week'].value_counts().plot.bar()'
```

```python

df['orders_day_of_week'].value_counts().plot.bar()
```

 On the left, you have your basic indexing syntax 
 Then after the dot comes the function you want to execute—in this case, value_counts(). 
  At the end comes the new, additional function: plot.bar()

```python
df['orders_day_of_week'].value_counts().plot.bar(color=['purple','red','pink','orange','yellow','green','blue'])
```

add (color=['']) to decide colors

```python
df['orders_day_of_week'].value_counts().sort_index().plot.bar()
```
